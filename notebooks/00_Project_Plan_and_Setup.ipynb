# 00 · Where Does the Software End? Project plan, roadmap and setup

**Aslan S. Dizaji · Kempner Institute Research Fellowship proposal → implementation**

The question: in one whole nervous system (*C. elegans*), what is the lowest level of description at which activity becomes self-contained "software", if such a level exists?

| Claim | Statement | Tested by |
|---|---|---|
| C1 simulability | dynamics approximable by a computable model | assumed (physical Church–Turing); B separates C1 from C2 |
| C2 computational closure | some level above biophysics is self-contained and substrate-invariant | E1 (closure ladder), E2 (substrate transfer) |
| C3 computer metaphor | the closed level is a program on hardware it does not rewrite | E3 (program vs hardware), E4 (embodiment) |

### Results so far (`laptop` mode, 2026-10-08)

| | finding |
|---|---|
| **Theory (01)** | Closure can hold observationally but fail causally; only interventions detect it. |
| **Data (03)** | 4.8% of tested pairs respond significantly; 29% of those have a direct contact. Split-half reliability is 0.42 over all pairs and 0.60 over significant pairs. The unc-31 peptide effect is not distinguishable from zero (SD ratio 0.62 [0.00, 0.99]). |
| **Validation (02)** | Errors are one-sided: the pipeline never picks too deep a rung, but misses L2 entirely and misses L3 unless the peptide effect matches wired signalling. Four times more trials do not help. |
| **E1 (04)** | Registered L* = L1, but all mechanistic rungs tie at ≈0.30 FEVE, the wiring-only L0 included. They underfit (training loss 0.81–0.92 vs noise floor 0.51), and longer training lowers held-out accuracy. The black box overfits (−0.46). No peptidergic signal. The least-explained responders are pharyngeal and gas-sensing neurons. |
| **E1b (06)** | Analyzing time courses instead of window means does not make L2/L3 detectable (no rescue). The window "hit" for L3 came from wide confidence intervals, not a measurable gain. |
| **E1c (07)** | The gap to the oracle is identifiability, not optimization: fits reach the truth's training loss but not its held-out accuracy. About 48 of ~900 parameter directions are stiff; the deeper mechanisms' parameters are essentially unconstrained. Averaging equally good fits does not help: they disagree about unseen interventions (ensemble fails the registered rule). On real E1, the necessity bracket is **L0 ≤ closed ≤ L1**. |
| **E1d (08)** | A 5-second pulse is the most informative new experiment for every model (real L1: 835 bits and 61 constrained directions, vs 176 and 24 for more of the same). But deeper mechanisms add only 0.008–0.019 FEVE, and the sufficiency (5%) and necessity (>0) rules are both miscalibrated for effects this small. |
| **E1e (09)** | A null-calibrated single threshold still goes too deep (3/6 and 1/6 test worms): the L2 rung's gain is the same (+0.001 to +0.004) whether or not the worm has adaptation, because it compensates for misfit. L3 with the long pulse was detected once (+0.045). On real data the closed level is L0 for every δ ≥ 0. |
| **E3 (10)** | Across 68 spontaneous recordings, a program fitted on the past goes stale (65/68 beyond stationary surrogates); drift exceeds threshold in 48/68 and no compact activity-driven meta-program absorbs it (median absorption −0.04). Verdicts: 47 rewired, 11 stationary, 10 underpowered, 0 self-programming. |
| **E3b (11)** | The drift is diffuse. It is not larger in the neurons whose behavioral encoding changed (median difference −0.0009 R², Wilcoxon p = 0.85). Slow behavioral state does not absorb it beyond a time-shifted control (p = 0.63): adding slow inputs, real or shifted, slightly *increases* the gap (0.0091 → 0.0106 / 0.0117). |
| **E3c (12)** | The drift looks like slow, independent changes of each neuron's drive, about twice the size of the planted drift. Gains and low-rank coupling changes add nothing. `dense` passes (p = 0.004), but just as strongly on drive-only surrogates, so R1 is invalid. The drive changes are not more shared than independent drift produces: PC1 0.29 vs 0.31 for drive-drift surrogates (the registered stationary null, 0.23, was the wrong comparison). Not the microscope: measurement corrections hurt, and drift does not grow with signal fading. |
| **E2 (05)** | The fitted L1 is invariant across Euler/RK4/Brian2 (gap ≈ 0.001 of noise) and has a negative Lyapunov exponent. A neuromorphic port needs 8-bit weights but 12-bit state. Weights are reproducible across fits; time constants are not. |

**Current reading:** at the resolution of window-averaged responses, nothing beyond wiring and signs is *detectably* needed, and the fitted description is a substrate-independent program. But that description explains only ~30% of explainable variance, and the test cannot see timing-based mechanisms. On amplitude data from single-neuron stimulation, the closed level is bracketed between L0 and L1. The deeper mechanisms are not *excluded*; they are *unidentifiable* from this kind of experiment (Notebooks 06–07).

### The ladder (what each rung adds)

| Rung | Mechanism | State / neuron | Notes |
|---|---|---|---|
| L0 | linear rate network on the connectome; signs; 5 global scales | 1 | "wiring and signs only" |
| L1 | graded nonlinear neurons; per-synapse weights with synaptic dynamics; gap-junction conductances | 2 | chemical + electrical |
| L2 | + slow K⁺ adaptation and regenerative inward current per neuron | 3 | intrinsic excitability |
| L3 | + slow, diffuse peptide release acting through the peptidergic connectome (switched off in unc-31) | 4 | extrasynaptic |
| L4 | + homeostatic integral feedback (activity-dependent slow state) | 5 | self-modification |
| B | transformer over neuron tokens | — | prediction without mechanism |

This ladder is a fast **dimensionless surrogate** written directly in JAX, so that E1's statistics can be developed and validated now. The proposal's version writes each rung once in NeuroML and compiles it to Brian2 or Jaxley; that port belongs to Notebook 05 (E2), where substrate is the variable under test. Your OpenWorm–Jaxley simulator is the starting point for it.

### Design decisions made during implementation (worth reflecting in the proposal)

1. **Primary E1 protocol = held-out (responder, stimulus) pairs, with a per-stimulus efficacy parameter.** Holding out whole stimulated neurons made every rung predict worse than zero on synthetic worms, because the efficacy of an unseen stimulation (opsin expression, own excitability) is not identifiable. Whole-stimulus holdout is kept as a stricter secondary test.
2. **Decision rule hardened:** L* must also predict significantly better than zero, so that a ladder of uniformly bad models cannot "close".
3. **Nested warm starts + equal optimization budget + restarts of L1.** Warm starts alone made deeper rungs win by inheriting extra training. On synthetic worms, truth L1 was recovered as L3. With equal budgets and 3 restarts it is recovered correctly. On the 60-neuron real wiring (Notebook 02, `laptop`), errors are one-sided: never too deep, often too shallow.
4. **Bilateral tying** of parameters (203 classes for 300 neurons).
5. **Artifact rejection:** trials with ΔF/F < −1 are physically impossible. There are only 11 of them, but they inflated noise variance eightfold.
6. **Power caveat:** after cleaning, the unc-31 peptidergic excess is not significantly different from zero with amplitude data alone (Notebook 03 §4). Notebook 02 §3 quantifies power for L2 vs L3.
