The question: in one whole nervous system (*C. elegans*), what is the lowest level of description at which activity becomes self-contained "software", if such a level exists?

| Claim | Statement | Tested by |
|---|---|---|
| C1 simulability | dynamics approximable by a computable model | assumed (physical Church–Turing); B separates C1 from C2 |
| C2 computational closure | some level above biophysics is self-contained and substrate-invariant | E1 (closure ladder), E2 (substrate transfer) |
| C3 computer metaphor | the closed level is a program on hardware it does not rewrite | E3 (program vs hardware), E4 (embodiment) |

### The ladder (what each rung adds)

| Rung | Mechanism | State / neuron | Notes |
|---|---|---|---|
| L0 | linear rate network on the connectome; signs; 5 global scales | 1 | "wiring and signs only" |
| L1 | graded nonlinear neurons; per-synapse weights with synaptic dynamics; gap-junction conductances | 2 | chemical + electrical |
| L2 | + slow K⁺ adaptation and regenerative inward current per neuron | 3 | intrinsic excitability |
| L3 | + slow, diffuse peptide release acting through the peptidergic connectome (switched off in unc-31) | 4 | extrasynaptic |
| L4 | + homeostatic integral feedback (activity-dependent slow state) | 5 | self-modification |
| B | transformer over neuron tokens | — | prediction without mechanism |

This ladder is a fast **dimensionless surrogate** written directly in JAX, so that E1's statistics can be developed and validated now. The proposal's version writes each rung once in NeuroML and compiles it to Brian2 or Jaxley; that port belongs to Notebook 05 (E2), where substrate is the variable under test. Your OpenWorm–Jaxley simulator is the starting point for it.

### Design decisions made during implementation (worth reflecting in the proposal)

1. **Primary E1 protocol = held-out (responder, stimulus) pairs, with a per-stimulus efficacy parameter.** Holding out whole stimulated neurons made every rung predict worse than zero on synthetic worms, because the efficacy of an unseen stimulation (opsin expression, own excitability) is not identifiable. Whole-stimulus holdout is kept as a stricter secondary test.
2. **Decision rule hardened:** L* must also predict significantly better than zero, so that a ladder of uniformly bad models cannot "close".
3. **Nested warm starts + equal optimization budget + restarts of L1.** Warm starts alone made deeper rungs win by inheriting extra training. On synthetic worms, truth L1 was recovered as L3. With equal budgets and 3 restarts it is recovered correctly. On the 60-neuron real wiring (Notebook 02, `laptop`), errors are one-sided: never too deep, often too shallow.
4. **Bilateral tying** of parameters (203 classes for 300 neurons).
5. **Artifact rejection:** trials with ΔF/F < −1 are physically impossible. There are only 11 of them, but they inflated noise variance eightfold.
6. **Power caveat:** after cleaning, the unc-31 peptidergic excess is not significantly different from zero with amplitude data alone (Notebook 03 §4). Notebook 02 §3 quantifies power for L2 vs L3.
